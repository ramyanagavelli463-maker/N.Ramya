```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns

from sklearn.cluster import KMeans
from sklearn.preprocessing import StandardScaler
```

```python
data = pd.read_csv('/content/customer_data (1).csv')
```

```python
data.head()
```

```python
data.info()
```

```python
data.isnull().sum()
```

```python
X = data[['Annual Income (k$)', 'Spending Score (1-100)']]
```

```python
X = data[['Annual_Income', 'Total_Amount_Spent']]

X.head()
```

```python
from sklearn.preprocessing import StandardScaler

scaler = StandardScaler()

X_scaled = scaler.fit_transform(X)
```

```python
X_scaled[:5]
```

```python
from sklearn.cluster import KMeans
import matplotlib.pyplot as plt

wcss = []

for i in range(1, 11):
    kmeans = KMeans(n_clusters=i, random_state=42, n_init=10)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.figure(figsize=(8,5))

plt.plot(range(1, 11), wcss, marker='o')

plt.title('Elbow Method')
plt.xlabel('Number of Clusters')
plt.ylabel('WCSS')

plt.show()
```

```python
kmeans = KMeans(n_clusters=4, random_state=42, n_init=10)

data['Cluster'] = kmeans.fit_predict(X_scaled)
```

```python
data.isnull().sum()
plt.figure(figsize=(10,6))

plt.scatter(
    data['Annual_Income'],
    data['Total_Amount_Spent'],
    c=data['Cluster'],
    s=50
)

plt.xlabel('Annual Income')
plt.ylabel('Total Amount Spent')
plt.title('Customer Segmentation')

plt.show()
```

```python
centers = scaler.inverse_transform(kmeans.cluster_centers_)

plt.figure(figsize=(10,6))

plt.scatter(
    data['Annual_Income'],
    data['Total_Amount_Spent'],
    c=data['Cluster'],
    s=50
)

plt.scatter(
    centers[:, 0],
    centers[:, 1],
    s=200,
    marker='X'
)

plt.xlabel('Annual Income')
plt.ylabel('Total Amount Spent')
plt.title('Customer Segmentation with Cluster Centers')

plt.show()
```

```python
cluster_summary = data.groupby('Cluster').agg({
    'Annual_Income': 'mean',
    'Total_Amount_Spent': 'mean',
    'Customer_Age': 'mean',
    'Family_Size': 'mean',
    'Web_Purchases': 'mean',
    'Store_Purchases': 'mean',
    'Catalog_Purchases': 'mean'
}).round(2)

cluster_summary
```
